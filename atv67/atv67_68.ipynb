```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'data': ['2023-01-15', '2023-01-20', '2023-02-10', '2023-02-15', '2023-03-05', '2023-03-22', '2023-04-12', '2023-04-18', '2023-05-01', '2023-05-25'],
    'loja': ['Loja A', 'Loja B', 'Loja A', 'Loja C', 'Loja B', 'Loja A', 'Loja C', 'Loja B', 'Loja A', 'Loja C'],
    'regiao': ['Sul', 'Norte', 'Sul', 'Leste', 'Norte', 'Sul', 'Leste', 'Norte', 'Sul', 'Leste'],
    'categoria': ['Eletrônicos', 'Roupas', 'Eletrônicos', 'Alimentos', 'Roupas', 'Móveis', 'Alimentos', 'Roupas', 'Eletrônicos', 'Móveis'],
    'produto': ['Notebook', 'Camiseta', 'Smartphone', 'Arroz', 'Calça', 'Cadeira', 'Feijão', 'Camiseta', 'Notebook', 'Mesa'],
    'quantidade': [5, 12, np.nan, 20, 15, 3, 25, 10, 8, np.nan], # Inclui valores nulos para testar
    'preco_unitario': [3500.0, 80.0, 2000.0, np.nan, 150.0, 450.0, 10.0, 80.0, 3500.0, 1200.0] # Inclui valores nulos
}
```

```python
df = pd.DataFrame(data)
```

```python
df['data'] = pd.to_datetime(df['data'])
```

```python
df['faturamento'] = df['quantidade'] * df['preco_unitario']
```

```python
print("--- 4a) Informações do Dataset (.info()) ---")[cite: 1]
df.info()[cite: 1]

print("\n--- 4b) Valores Ausentes (.isnull().sum()) ---")[cite: 1]
print(df.isnull().sum())[cite: 1]

print("\n--- 4c) Resumo Estatístico (.describe()) ---")[cite: 1]
print(df.describe())[cite: 1]
```

```python
df['quantidade'] = df['quantidade'].fillna(df['quantidade'].mean())[cite: 1]
df['preco_unitario'] = df['preco_unitario'].fillna(df['preco_unitario'].mean())[cite: 1]
```

```python
df['faturamento'] = df['quantidade'] * df['preco_unitario']
```

```python
df['mes'] = df['data'].dt.month[cite: 1]
df['mes_ano'] = df['data'].dt.strftime('%m/%Y')[cite: 1]
```

```python
print("\n--- 6) Faturamento Total por Loja ---")[cite: 1]
faturamento_por_loja = df.groupby('loja')['faturamento'].sum().sort_values(ascending=False)[cite: 1]
print(faturamento_por_loja)
```

```python
print("\n--- 7) Ticket Médio por Loja ---")[cite: 1]
ticket_medio = df.groupby('loja')['faturamento'].sum() / df.groupby('loja')['faturamento'].count()[cite: 1]
print(ticket_medio)
```

```python
print("\n--- 8) Top 5 Produtos Mais Vendidos (Quantidade) ---")[cite: 1]
top5_produtos = df.groupby('produto')['quantidade'].sum().sort_values(ascending=False).head(5)[cite: 1]
print(top5_produtos)
```

```python
faturamento_mensal = df.groupby('mes_ano')['faturamento'].sum()[cite: 1]
mes_maior = faturamento_mensal.idxmax()[cite: 1]
mes_menor = faturamento_mensal.idxmin()[cite: 1]
print(f"\n--- 9) Maior Faturamento: {mes_maior} | Menor Faturamento: {mes_menor} ---")[cite: 1]
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x=faturamento_por_loja.index, y=faturamento_por_loja.values)
plt.title('Faturamento Total por Loja', fontsize=14)[cite: 2]
plt.xlabel('Loja', fontsize=12)[cite: 2]
plt.ylabel('Faturamento (R$)', fontsize=12)[cite: 2]
plt.tight_layout()
plt.savefig('graficos/faturamento_por_loja.png')[cite: 2]
plt.show()
```

```python
plt.figure(figsize=(8, 5))
faturamento_por_mes = df.groupby('mes_ano')['faturamento'].sum()
plt.plot(faturamento_por_mes.index, faturamento_por_mes.values, marker='o', linewidth=2)[cite: 2]
for i, valor in enumerate(faturamento_por_mes.values):
    plt.text(i, valor, f'R${valor:,.0f}', ha='center', va='bottom')[cite: 2]
plt.title('Evolução Mensal do Faturamento', fontsize=14)[cite: 2]
plt.xlabel('Mês/Ano', fontsize=12)[cite: 2]
plt.ylabel('Faturamento (R$)', fontsize=12)[cite: 2]
plt.tight_layout()
plt.savefig('graficos/evolucao_faturamento.png')[cite: 2]
plt.show()
```
